## 5. Rules for Naming Variables (Identifiers)

Variable, function, class ইত্যাদির নামকে identifier বলা হয়।

### Naming rules
1. নামের মধ্যে letter, digit এবং underscore (`_`) থাকতে পারে।
2. নাম digit দিয়ে শুরু হতে পারে না।
3. নামের মধ্যে space দেওয়া যায় না।
4. Python keyword variable name হিসেবে ব্যবহার করা যায় না।
5. Python case-sensitive।

### Valid names
- `student_name`
- `age2`
- `_total`
- `final_score`

### Invalid names
- `2age` — digit দিয়ে শুরু
- `student name` — space আছে
- `class` — Python keyword

## 9. Operators in Python

Operator হলো symbol বা keyword, যা value/variable-এর ওপর operation সম্পন্ন করে।

এই Part-এ শিখব:
1. Arithmetic operators
2. Comparison operators
3. Assignment operators
4. Logical operators
5. Membership ও identity operators-এর পরিচিতি

### 9.1 Arithmetic operators

| Operator | কাজ | উদাহরণ |
|---|---|---|
| `+` | যোগ | `10 + 3` |
| `-` | বিয়োগ | `10 - 3` |
| `*` | গুণ | `10 * 3` |
| `/` | ভাগ; ফল float | `10 / 3` |
| `//` | floor division | `10 // 3` |
| `%` | ভাগশেষ | `10 % 3` |
| `**` | power | `2 ** 3` |

```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

### 9.2 Comparison operators

Comparison-এর result `True` অথবা `False` হয়।

```python
x = 10
y = 20

print(x == y)  # Equal to
print(x != y)  # Not equal to
print(x > y)   # Greater than
print(x < y)   # Less than
print(x >= 10) # Greater than or equal
print(y <= 20) # Less than or equal
```

### 9.3 Assignment operators

Assignment operator variable-এ value assign বা update করে।

```python
score = 10
print(score)

score += 5   # score = score + 5
print(score)

score *= 2   # score = score * 2
print(score)

score -= 4
print(score)
```

### 9.4 Logical operators

- `and`: সব condition True হলে True
- `or`: অন্তত একটি condition True হলে True
- `not`: Boolean result উল্টে দেয়

```python
age = 22
has_id = True

print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
```

### 9.5 Membership operators

`in` এবং `not in` দিয়ে কোনো item কোনো sequence-এর মধ্যে আছে কি না পরীক্ষা করা যায়।

```python
course = "Python Programming"

print("Python" in course)
print("Java" in course)
print("Java" not in course)
```

### 9.6 Identity operators

`is` এবং `is not` পরীক্ষা করে দুটি reference একই object নির্দেশ করছে কি না। এটি `==`-এর মতো নয়; `==` value সমান কি না পরীক্ষা করে।

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a == c)  # Values are equal
print(a is c)  # They are different list objects
print(a is b)  # b refers to the same object as a
```

## 10. Operator Precedence

একটি expression-এ একাধিক operator থাকলে কোন operation আগে হবে, তা operator precedence নির্ধারণ করে।

সাধারণভাবে:
1. Parentheses `()`
2. Exponentiation `**`
3. Multiplication, division, floor division, modulus
4. Addition, subtraction
5. Comparisons
6. `not`
7. `and`
8. `or`

জটিল expression-এ parentheses ব্যবহার করলে code বেশি পরিষ্কার হয়।

```python
result1 = 10 + 2 * 3
result2 = (10 + 2) * 3

print(result1)
print(result2)
```

## 11. Useful Built-in Functions

কিছু built-in function শুরু থেকেই Python-এ থাকে; আলাদা করে install করতে হয় না।

- `print()` — output দেখায়

## 15. Common Beginner Errors

| Error | সাধারণ কারণ | কী করবেন |
|---|---|---|
| `SyntaxError` | Python syntax ভুল | quotation, bracket, colon পরীক্ষা করুন |
| `NameError` | variable define না করে ব্যবহার | spelling ও আগে assign করা হয়েছে কি না দেখুন |
| `TypeError` | incompatible data type-এর operation | type পরীক্ষা করুন, প্রয়োজন হলে conversion করুন |
| `ValueError` | সঠিক type-এ convert করা যাচ্ছে না | input value valid কি না দেখুন |
| `IndentationError` | indentation ভুল | block-এর ভেতরে একই indentation রাখুন |

### Error বুঝতে শেখা
Error message-এর শেষের অংশে error type এবং সাধারণত কোন line-এ সমস্যা হয়েছে তা থাকে। Error দেখলে ভয় পাবেন না—ধাপে ধাপে line পরীক্ষা করুন।
