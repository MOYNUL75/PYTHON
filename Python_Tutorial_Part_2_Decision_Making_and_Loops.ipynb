## 1. What is Control Flow?

Control flow হলো program-এর কোন instruction কখন এবং কোন ক্রমে execute হবে তা নিয়ন্ত্রণ করার পদ্ধতি।

**দুইটি গুরুত্বপূর্ণ control structure:**
- **Decision making:** condition অনুযায়ী কোন code চলবে, যেমন `if`, `else`, `elif`।
- **Looping:** কোনো code বারবার চালানো, যেমন `for` ও `while`।

## 2. Conditions and Boolean Results

Condition এমন একটি expression যার ফল `True` অথবা `False` হয়। Decision making-এর ভিত্তি হলো condition।

Comparison operators:
| Operator | অর্থ |
|---|---|
| `==` | সমান |
| `!=` | সমান নয় |
| `>` | বড় |
| `<` | ছোট |
| `>=` | বড় অথবা সমান |
| `<=` | ছোট অথবা সমান |

````python
age = 20

print(age == 20)
print(age != 20)
print(age > 18)
print(age < 18)
print(age >= 20)
print(age <= 20)
````

### 2.1 Combining conditions with logical operators

- `and`: সব condition True হতে হবে।
- `or`: অন্তত একটি condition True হলেই হবে।
- `not`: Boolean result উল্টে দেয়।

````python
age = 22
has_student_id = True

print(age >= 18 and has_student_id)
print(age < 18 or has_student_id)
print(not has_student_id)
````

## 3. The `if` Statement

`if` statement কোনো condition True হলে নির্দিষ্ট code execute করে।

**Syntax**
```python
if condition:
    statement
```

মনে রাখবে:
- `if`-এর পরে condition লিখতে হয়।
- condition-এর পরে colon `:` দিতে হয়।
- condition True হলে indented block execute হয়।

````python
temperature = 32

if temperature > 30:
    print("It is a hot day.")
````

### 3.1 `if` condition False হলে কী হয়?
Condition False হলে `if` block-এর ভেতরের code execute হয় না। Program পরের statement-এ চলে যায়।

````python
temperature = 25

if temperature > 30:
    print("It is a hot day.")

print("Program continues.")
````

## 4. The `if-else` Statement

যখন condition True হলে এক কাজ এবং False হলে অন্য কাজ করতে হয়, তখন `if-else` ব্যবহার করা হয়।


## 5. The `if-elif-else` Statement

একাধিক condition পরীক্ষা করতে `elif` ব্যবহার করা হয়। Python ওপর থেকে নিচে condition পরীক্ষা করে; প্রথম True condition-এর block execute করে, তারপর পুরো structure থেকে বের হয়ে যায়।

**Syntax**
```python
if condition_1:
    statement_1
elif condition_2:
    statement_2
elif condition_3:
    statement_3
else:
    default_statement
```

````python
marks = 85

if marks >= 80:
    print("Grade: A+")
elif marks >= 70:
    print("Grade: A")
elif marks >= 60:
    print("Grade: A-")
elif marks >= 50:
    print("Grade: B")
elif marks >= 40:
    print("Grade: C")
else:
    print("Grade: F")
````

## 8. The `for` Loop

`for` loop কোনো sequence বা iterable-এর প্রতিটি item-এর জন্য code execute করে।

**Syntax**
```python
for variable in sequence:
    statements
```

এখানে `variable` প্রতিবার sequence-এর একটি করে value পায়।

````python
subjects = ["Python", "Database", "Networking"]

for subject in subjects:
    print(subject)
````

## 9. The `range()` Function

`range()` সাধারণত নির্দিষ্ট সংখ্যক iteration চালাতে ব্যবহৃত হয়।

তিনটি প্রচলিত form:
- `range(stop)` — 0 থেকে `stop - 1`
- `range(start, stop)` — `start` থেকে `stop - 1`
- `range(start, stop, step)` — নির্দিষ্ট step-এ সংখ্যা তৈরি করে

**খুব গুরুত্বপূর্ণ:** `stop` value অন্তর্ভুক্ত হয় না।

````python
# 0 to 4
for i in range(5):
    print(i)
````

````python
# 1 to 5
for i in range(1, 6):
    print(i)
````

````python
# Even numbers from 2 to 10
for i in range(2, 11, 2):
    print(i)
````

````python
# Countdown: 5 to 1
for i in range(5, 0, -1):
    print(i)
````

## 10. The `while` Loop

`while` loop condition True থাকা পর্যন্ত code execute করে।

**Syntax**
```python
while condition:
    statements
```

সাধারণত loop-এর ভেতরে এমনভাবে variable update করতে হয়, যাতে একসময় condition False হয়।

````python
count = 1

while count <= 5:
    print(count)
    count = count + 1
````

### 10.1 Trace the loop step by step

প্রতিবার `count`-এর value কী হচ্ছে তা লক্ষ্য করো।

````python
count = 1

while count <= 4:
    print("Current count:", count)
    count += 1

print("Loop finished.")
````

### 10.2 Sum using a `while` loop

````python
number = 1
total = 0

while number <= 10:
    total += number
    number += 1

print("Sum:", total)
````

### 10.3 Important: Infinite loop

যদি condition কখনো False না হয়, loop চলতেই থাকবে। যেমন `count` update না করলে নিচের loop বন্ধ হবে না।

এটি run কোরো না। শুধু code দেখে সমস্যাটি বোঝার চেষ্টা করো।
```python
count = 1
while count <= 5:
    print(count)
    # count update করা হয়নি
```

Jupyter-এ ভুল করে infinite loop চালু হলে Kernel → Interrupt অথবা toolbar-এর interrupt button ব্যবহার করো।

## 11. `break` Statement

`break` loop-কে সঙ্গে সঙ্গে বন্ধ করে দেয়। Condition True হলেই loop থেকে বের হয়ে যায়।

````python
for number in range(1, 10):
    if number == 5:
        break
    print(number)

print("Loop stopped.")
````

### 11.1 Search for a value

একটি list-এর মধ্যে target value খুঁজে পেলে search বন্ধ করা যায়।

````python
numbers = [12, 25, 37, 48, 59]
target = 37

for number in numbers:
    print("Checking:", number)

    if number == target:
        print("Target found!")
        break
````

## 12. `continue` Statement

`continue` বর্তমান iteration-এর বাকি code বাদ দিয়ে পরের iteration শুরু করে।

````python
for number in range(1, 6):
    if number == 3:
        continue

    print(number)
````

### 12.1 Print only odd numbers

`pass` কোনো কাজ করে না; এটি placeholder হিসেবে ব্যবহৃত হয়। কোনো block আপাতত খালি রাখতে হলে এটি কাজে লাগে।

## 14. Nested Loops

একটি loop-এর ভেতরে আরেকটি loop থাকলে তাকে nested loop বলে।

Outer loop-এর প্রতিটি iteration-এর জন্য inner loop সম্পূর্ণ চলে।

````python
for row in range(1, 4):
    for column in range(1, 4):
        print("Row:", row, "Column:", column)
````

### 14.1 Print a simple star pattern

### 14.2 Multiplication tables from 1 to 3

````python
for number in range(1, 4):
    print("Multiplication table of", number)

    for multiplier in range(1, 6):
        print(number, "x", multiplier, "=", number * multiplier)

    print()
````

## 15. `for` Loop vs `while` Loop

| বিষয় | `for` loop | `while` loop |
|---|---|---|
| ব্যবহার | sequence/iterable-এর item নিয়ে কাজ | condition True থাকা পর্যন্ত |
| Iteration | sequence অনুযায়ী | condition ও update-এর ওপর নির্ভর |
| Counter update | অনেক ক্ষেত্রে স্বয়ংক্রিয়ভাবে item বদলায় | প্রয়োজন হলে নিজে update করতে হয় |
| সাধারণ উদাহরণ | list-এর item print | user সঠিক input না দেওয়া পর্যন্ত |

দুটিই repetition-এর জন্য। সমস্যার ধরন অনুযায়ী loop বেছে নাও।

## 17. Mini Project 2 — Simple Login Attempts

User-কে সর্বোচ্চ 3 বার password দেওয়ার সুযোগ দেওয়া হবে। সঠিক password দিলে loop বন্ধ হবে।

````python
correct_password = "python123"
attempts = 0
max_attempts = 3

while attempts < max_attempts:
    password = input("Enter password: ")
    attempts += 1

    if password == correct_password:
        print("Login successful.")
        break
    else:
        remaining = max_attempts - attempts
        print("Incorrect password.")

        if remaining > 0:
            print("Attempts remaining:", remaining)
else:
    print("Account access denied for this session.")
````

## 18. Mini Project 3 — Number Guessing Game

এখানে computer-এর নির্ধারিত number অনুমান করতে হবে। `break` সঠিক উত্তর পেলে game বন্ধ করবে।

````python
secret_number = 7

while True:
    guess = int(input("Guess the number (1-10): "))

    if guess == secret_number:
        print("Correct! You won.")
        break
    elif guess < secret_number:
        print("Too low. Try again.")
    else:
        print("Too high. Try again.")
````

## 20. Common Errors and Troubleshooting

| Error / Problem | কারণ | সমাধান |
|---|---|---|
| `SyntaxError` | colon বা syntax ভুল | `if`, `else`, `for`, `while` line পরীক্ষা |
| `IndentationError` | block-এর indentation ভুল | একই block-এ consistent 4 spaces |
| Infinite loop | `while` condition কখনো False হচ্ছে না | loop variable update করো |
| Wrong grade | condition ভুল order-এ লেখা | বড় threshold আগে পরীক্ষা করো |
| Loop runs one extra/fewer time | `range()`-এর stop exclusive | `range(start, stop)`-এ stop অন্তর্ভুক্ত নয় |
| `ValueError` | numeric input-এ text দেওয়া | valid number input দাও |

- `break` loop বন্ধ করে।
- `continue` বর্তমান iteration skip করে।
