# Section A — Classes and Objects

## 1. Class কী?

Class হলো object তৈরির blueprint বা template। Class-এ attributes এবং methods define করা হয়।

**Syntax**
```python
class ClassName:
    # attributes and methods
    pass
```

`pass` হলো placeholder; পরে code যোগ করা যাবে।

## 2. Object কী?

Object হলো class থেকে তৈরি instance। একই class থেকে অনেক object তৈরি করা যায়; প্রতিটি object-এর data আলাদা হতে পারে।

````python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1)
print(student2)
print(type(student1))
````

## 3. Attributes যোগ করা

Object-এর attribute dot notation দিয়ে set বা access করা যায়। তবে বড় program-এ attributes initialize করার জন্য সাধারণত `__init__()` ব্যবহার করা হয়।

````python
class Student:
    pass

student1 = Student()
student1.name = "Amina"
student1.department = "CSE"
student1.cgpa = 3.80

print(student1.name)
print(student1.department)
print(student1.cgpa)
````

## 4. `__init__()` Constructor

`__init__()` হলো initializer method। Object তৈরি হওয়ার সময় Python এটি automatically call করে এবং object-এর initial state set করতে ব্যবহার করা হয়।

`self` current instance-কে নির্দেশ করে। Instance method-এর প্রথম parameter হিসেবে convention অনুযায়ী `self` লেখা হয়।

````python
class Student:
    def __init__(self, name, department, cgpa):
        self.name = name
        self.department = department
        self.cgpa = cgpa

student1 = Student("Amina", "CSE", 3.80)
student2 = Student("Rahim", "EEE", 3.65)

print(student1.name, student1.department, student1.cgpa)
print(student2.name, student2.department, student2.cgpa)
````

## 5. Instance Methods

Class-এর ভেতরে define করা function-কে method বলা হয়। Instance method object-এর data নিয়ে কাজ করে এবং সাধারণত `self` গ্রহণ করে।

````python
class Student:
    def __init__(self, name, cgpa):
        self.name = name
        self.cgpa = cgpa

    def introduce(self):
        print(f"My name is {self.name}.")

    def show_cgpa(self):
        print(f"My CGPA is {self.cgpa:.2f}")

student = Student("Nila", 3.90)
student.introduce()
student.show_cgpa()
````

## 6. Updating Object Attributes

Instance attribute update করতে method অথবা dot notation ব্যবহার করা যায়। Encapsulation শেখার পর controlled update-এর ধারণা আরও পরিষ্কার হবে।

````python
class Student:
    def __init__(self, name, cgpa):
        self.name = name
        self.cgpa = cgpa

    def update_cgpa(self, new_cgpa):
        self.cgpa = new_cgpa

student = Student("Hasan", 3.40)
print("Before:", student.cgpa)

student.update_cgpa(3.60)
print("After:", student.cgpa)
````

# Section B — Attributes and Methods in More Detail

## 7. Instance Attributes vs Class Attributes

- **Instance attribute:** প্রতিটি object-এর নিজস্ব value থাকে।
- **Class attribute:** class-এর সঙ্গে যুক্ত; instance-এ আলাদা করে override না করলে shared class value পাওয়া যায়।

উদাহরণ: সব student-এর institution একই হতে পারে, কিন্তু name আলাদা।

````python
class Student:
    institution = "ABC University"  # Class attribute

    def __init__(self, name, department):
        self.name = name            # Instance attribute
        self.department = department

student1 = Student("Amina", "CSE")
student2 = Student("Rahim", "BBA")

print(student1.institution)
print(student2.institution)
print(student1.name)
print(student2.name)
````

### 7.1 Updating a class attribute

Class-এর মাধ্যমে class attribute update করলে shared value পরিবর্তিত হয়। কোনো instance-এ একই নাম assign করলে সেটি instance attribute তৈরি করে class attribute-কে shadow করতে পারে।

````python
class Employee:
    company = "Tech Solutions"

    def __init__(self, name):
        self.name = name

emp1 = Employee("Rafi")
emp2 = Employee("Mina")

Employee.company = "NextGen Ltd."

print(emp1.company)
print(emp2.company)

emp1.company = "Personal Company"
print(emp1.company)
print(emp2.company)
print(Employee.company)
````

## 8. Instance Methods, Class Methods and Static Methods

তিন ধরনের method-এর সাধারণ ব্যবহার:
- Instance method: object-এর data নিয়ে কাজ করে; `self` নেয়।
- Class method: class-level data নিয়ে কাজ করে; `@classmethod` এবং `cls` ব্যবহার করে।
- Static method: class-এর namespace-এ সম্পর্কিত utility function; `self` বা `cls` automatically নেয় না।

````python
class Student:
    university = "ABC University"

    def __init__(self, name):
        self.name = name

    def introduce(self):                 # Instance method
        return f"I am {self.name}."

    @classmethod
    def change_university(cls, new_name):
        cls.university = new_name

    @staticmethod
    def is_valid_cgpa(cgpa):
        return 0.0 <= cgpa <= 4.0

student = Student("Amina")

print(student.introduce())
print(Student.university)

Student.change_university("XYZ University")
print(Student.university)

print(Student.is_valid_cgpa(3.75))
print(Student.is_valid_cgpa(4.5))
````

# Section C — Encapsulation

## 9. Encapsulation কী?

Encapsulation হলো data ও সংশ্লিষ্ট methods একসঙ্গে রাখা এবং object-এর internal state কীভাবে পরিবর্তিত হবে তা নিয়ন্ত্রণ করার design principle।

Python-এ:
- `name` — public convention
- `_name` — protected/internal use-এর convention; language-level strict protection নয়
- `__name` — name mangling হয়; এটিও absolute privacy/security নয়

Python convention-কে গুরুত্ব দেয়; Java/C++-এর মতো strict private access control নেই।

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance  # Internal-use convention

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit must be positive.")
            return
        self._balance += amount

    def get_balance(self):
        return self._balance

account = BankAccount("Amina", 1000)
account.deposit(500)

print(account.owner)
print(account.get_balance())
````

## 10. Name Mangling with Double Underscore

`__balance`-এর মতো double underscore attribute class-এর ভেতরে name-mangled হয়। এটি accidental name collision কমাতে সাহায্য করে; security boundary নয়।

````python
class Example:
    def __init__(self):
        self.__secret = "Hidden by name mangling"

    def show_secret(self):
        return self.__secret

obj = Example()
print(obj.show_secret())

# Inspect names for learning:
print([name for name in dir(obj) if "secret" in name])
````

## 11. Properties — Controlled Attribute Access

`@property` দিয়ে method-কে attribute-এর মতো access করা যায়। Setter ব্যবহার করে validation করা যায়।

````python
class Product:
    def __init__(self, price):
        self.price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("Price cannot be negative.")
        self._price = value

product = Product(250)
print(product.price)

product.price = 300
print(product.price)
````

নিচের statement চালালে `ValueError` হবে, কারণ negative price গ্রহণ করা হয় না:
```python
product.price = -50
```

# Section D — Inheritance

## 12. Inheritance কী?

Inheritance-এর মাধ্যমে একটি class অন্য class-এর attributes ও methods inherit করতে পারে।

- Parent/base class: যে class থেকে inherit করা হয়।
- Child/derived class: যে class inherit করে।

এতে common behaviour reuse করা যায়।

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"My name is {self.name} and I am {self.age}."

class Student(Person):
    def study(self):
        return f"{self.name} is studying."

student = Student("Amina", 21)

print(student.introduce())
print(student.study())
````

## 13. `super()` Function

`super()` parent class-এর method বা initializer call করতে সাহায্য করে। Child class-এ নতুন attribute যোগ করার সময় parent initializer reuse করা যায়।

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id

student = Student("Rahim", 22, "CSE-101")

print(student.name)
print(student.age)
print(student.student_id)
````

## 14. Method Overriding

Child class parent class-এর method একই নামে define করলে child-এর implementation ব্যবহার হয়। একে method overriding বলে।

````python
class Animal:
    def sound(self):
        return "Some generic sound"

class Dog(Animal):
    def sound(self):
        return "Woof!"

class Cat(Animal):
    def sound(self):
        return "Meow!"

print(Animal().sound())
print(Dog().sound())
print(Cat().sound())
````

## 15. Types of Inheritance

Common inheritance structures:
1. Single inheritance — একটি parent class
2. Multiple inheritance — একাধিক parent class
3. Multilevel inheritance — inheritance chain
4. Hierarchical inheritance — একটি parent-এর একাধিক child
5. Hybrid inheritance — একাধিক structure-এর combination

শুরুতে single ও multilevel inheritance ভালোভাবে বুঝবে।

````python
# Single inheritance
class Vehicle:
    def move(self):
        return "Vehicle is moving"

class Car(Vehicle):
    pass

print(Car().move())
````

````python
# Multilevel inheritance
class LivingThing:
    def alive(self):
        return "This is alive"

class Animal(LivingThing):
    def breathe(self):
        return "Breathing"

class Bird(Animal):
    def fly(self):
        return "Flying"

bird = Bird()
print(bird.alive())
print(bird.breathe())
print(bird.fly())
````

````python
# Hierarchical inheritance
class Shape:
    def category(self):
        return "A shape"

class Circle(Shape):
    pass

class Rectangle(Shape):
    pass

print(Circle().category())
print(Rectangle().category())
````

````python
# Multiple inheritance
class Camera:
    def take_photo(self):
        return "Photo captured"

class Phone:
    def make_call(self):
        return "Calling"

class Smartphone(Camera, Phone):
    pass

device = Smartphone()
print(device.take_photo())
print(device.make_call())
````

### 15.1 Method Resolution Order (MRO)

Multiple inheritance-এ Python কোন class থেকে method নেবে তা MRO নির্ধারণ করে। `ClassName.mro()` দিয়ে order দেখা যায়।

````python
class A:
    def message(self):
        return "A"

class B(A):
    def message(self):
        return "B"

class C(A):
    def message(self):
        return "C"

class D(B, C):
    pass

print(D.mro())
print(D().message())
````

# Section E — Polymorphism and Abstraction

## 16. Polymorphism কী?

Polymorphism অর্থ একই interface বা method name বিভিন্ন object-এর ক্ষেত্রে ভিন্ন behaviour দিতে পারে।

উদাহরণ: `sound()` method Dog ও Cat-এর জন্য আলাদা output দেয়।

````python
class Dog:
    def sound(self):
        return "Woof!"

class Cat:
    def sound(self):
        return "Meow!"

class Cow:
    def sound(self):
        return "Moo!"

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(animal.sound())
````

## 17. Duck Typing

Python-এ object-এর exact class-এর চেয়ে object প্রয়োজনীয় method support করে কি না, তা গুরুত্বপূর্ণ হতে পারে। একে duck typing বলা হয়।

````python
class PDFReport:
    def export(self):
        return "Exporting PDF report"

class ExcelReport:
    def export(self):
        return "Exporting Excel report"

def export_report(report):
    print(report.export())

export_report(PDFReport())
export_report(ExcelReport())
````

## 18. Abstraction কী?

Abstraction হলো প্রয়োজনীয় interface প্রকাশ করা এবং implementation details আড়াল রাখা।

Python-এর `abc` module দিয়ে abstract base class তৈরি করা যায়। `@abstractmethod` দিয়ে child class-কে নির্দিষ্ট method implement করতে বলা যায়।

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

shapes = [Circle(3), Rectangle(4, 5)]

for shape in shapes:
    print("Area:", shape.area())
````

Abstract class-এর abstract method implement না করলে child class-এর object তৈরি করা যায় না। নিচের code-টি run কোরো না; এটি `TypeError` তৈরি করবে:

```python
class IncompleteShape(Shape):
    pass

obj = IncompleteShape()
```

# Section F — Composition and Special Methods

## 19. Composition

Composition হলো একটি class-এর object-এর ভেতরে অন্য class-এর object রাখা। এটি "has-a" relationship বোঝায়।

উদাহরণ: একটি Car-এর একটি Engine আছে।

````python
class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self, model):
        self.model = model
        self.engine = Engine()

    def start_car(self):
        return f"{self.model}: {self.engine.start()}"

car = Car("Toyota")
print(car.start_car())
````

## 20. Magic / Dunder Methods

Dunder মানে double underscore। Python-এর special methods built-in operations-এর behaviour define করতে সাহায্য করে।

কিছু পরিচিত method:
- `__init__()` — initialize object
- `__str__()` — readable string representation
- `__repr__()` — developer-oriented representation
- `__len__()` — `len()` behaviour
- `__eq__()` — equality comparison
- `__add__()` — `+` operator behaviour

````python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"{self.title} ({self.pages} pages)"

    def __len__(self):
        return self.pages

book = Book("Learning Python", 350)

print(book)
print(len(book))
````

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

p1 = Point(2, 3)
p2 = Point(4, 5)

print(p1 + p2)
````

## 21. Mini Project 1 — Bank Account System

এই project-এ class, constructor, methods, validation এবং encapsulation convention একসঙ্গে ব্যবহার করা হয়েছে।

````python
class BankAccount:
    def __init__(self, owner, initial_balance=0):
        if initial_balance < 0:
            raise ValueError("Initial balance cannot be negative.")

        self.owner = owner
        self._balance = initial_balance

    def deposit(self, amount):
        if amount <= 0:
            return "Deposit amount must be positive."
        self._balance += amount
        return f"Deposited {amount:.2f}"

    def withdraw(self, amount):
        if amount <= 0:
            return "Withdrawal amount must be positive."
        if amount > self._balance:
            return "Insufficient balance."
        self._balance -= amount
        return f"Withdrawn {amount:.2f}"

    def get_balance(self):
        return self._balance

account = BankAccount("Amina", 1000)

print(account.deposit(500))
print(account.withdraw(200))
print(account.withdraw(5000))
print("Current balance:", account.get_balance())
````

## 22. Mini Project 2 — Student Management

একটি Student class তৈরি করে student-এর information ও result দেখানো হবে।

````python
class Student:
    institution = "ABC University"

    def __init__(self, student_id, name, marks):
        self.student_id = student_id
        self.name = name
        self.marks = marks

    def average(self):
        if not self.marks:
            return 0
        return sum(self.marks) / len(self.marks)

    def result(self):
        avg = self.average()
        status = "Pass" if avg >= 40 else "Needs improvement"
        return avg, status

    def display(self):
        avg, status = self.result()
        print(f"ID: {self.student_id}")
        print(f"Name: {self.name}")
        print(f"Institution: {self.institution}")
        print(f"Average: {avg:.2f}")
        print(f"Status: {status}")

student = Student("CSE-101", "Rahim", [75, 82, 68, 90])
student.display()
````

## 23. Mini Project 3 — Employee and Manager Inheritance

Employee parent class থেকে Manager child class inherit করবে এবং অতিরিক্ত department attribute যোগ করবে।

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def describe(self):
        return f"{self.name} earns {self.salary:.2f}."

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def describe(self):
        return (
            f"{self.name} manages {self.department} "
            f"and earns {self.salary:.2f}."
        )

employee = Employee("Nila", 30000)
manager = Manager("Hasan", 50000, "IT")

print(employee.describe())
print(manager.describe())
````

### Solutions 1–4: Basic classes and methods

````python
# 1
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

car = Car("Toyota", "Corolla")
print(car.brand, car.model)

# 2
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

shape = Rectangle(10, 5)
print(shape.area(), shape.perimeter())

# 3
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

emp = Employee("Amina", 35000)
print(emp.name, emp.salary)

# 4
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount

account = BankAccount(1000)
account.deposit(500)
account.withdraw(200)
print(account.balance)
````

### Solutions 5–8: Class methods, static methods and properties

````python
# 5–6
class University:
    name = "ABC University"

    @classmethod
    def change_name(cls, new_name):
        cls.name = new_name

University.change_name("XYZ University")
print(University.name)

# 7
class NumberTools:
    @staticmethod
    def is_positive(number):
        return number > 0

print(NumberTools.is_positive(5))

# 8
class Product:
    def __init__(self, price):
        self.price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("Price cannot be negative")
        self._price = value

item = Product(100)
item.price = 150
print(item.price)
````

### Solutions 9–11: Inheritance and polymorphism

````python
# 9–10
class Person:
    def __init__(self, name):
        self.name = name

    def role(self):
        return "Person"

class Teacher(Person):
    def role(self):
        return "Teacher"

teacher = Teacher("Rina")
print(teacher.name, teacher.role())

# 11
class Dog:
    def sound(self):
        return "Woof"

class Cat:
    def sound(self):
        return "Meow"

class Cow:
    def sound(self):
        return "Moo"

for animal in [Dog(), Cat(), Cow()]:
    print(animal.sound())
````

### Solutions 12–14: Abstraction and dunder methods

````python
# 12
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CardPayment(Payment):
    def pay(self, amount):
        return f"Card payment: {amount}"

class CashPayment(Payment):
    def pay(self, amount):
        return f"Cash payment: {amount}"

print(CardPayment().pay(500))
print(CashPayment().pay(300))

# 13
class Book:
    def __init__(self, title):
        self.title = title

    def __str__(self):
        return f"Book: {self.title}"

print(Book("Python Basics"))

# 14
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

print(Point(1, 2) + Point(3, 4))
````

## 24. Common OOP Errors and Troubleshooting

| Error / Problem | কারণ | সমাধান |
|---|---|---|
| `TypeError: missing ... argument` | Constructor/method-এর required argument দেওয়া হয়নি | object creation ও method call পরীক্ষা |
| `AttributeError` | Attribute বা method নেই/নাম ভুল | spelling ও object type পরীক্ষা |
| `NameError: self is not defined` | Method-এ `self` parameter বাদ পড়েছে বা ভুলভাবে ব্যবহার | instance method-এ প্রথম parameter হিসেবে `self` দাও |
| Parent attribute missing | Child initializer-এ parent initializer call হয়নি | প্রয়োজন হলে `super().__init__(...)` ব্যবহার |
| Abstract class instantiate error | Abstract method implement হয়নি | child class-এ সব abstract method implement |
| Unexpected class attribute value | Instance attribute class attribute shadow করেছে | `ClassName.attribute` ও `instance.__dict__` পরীক্ষা |

### Important reminders
- Class blueprint; object হলো instance।
- `self` current object-কে নির্দেশ করে।
- `__init__()` initial state set করে।
- `_attribute` convention; strict privacy নয়।
- Inheritance হলো "is-a" relationship; composition সাধারণত "has-a"।
- Overriding child class-এ inherited method-এর implementation বদলায়।
- Polymorphism একই interface-এ different behaviour দিতে পারে।
