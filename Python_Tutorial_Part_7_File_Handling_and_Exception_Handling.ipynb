## 5. The `with open()` Context Manager

`with open()` ব্যবহার করলে block শেষ হওয়ার পর file automatically close হয়। এটি recommended practice।

````python
with open("part7_first_file.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
print("File is closed:", file.closed)
````

## 6. File Modes

| Mode | Meaning |
|---|---|
| `r` | Read; file থাকতে হবে |
| `w` | Write; নতুন file তৈরি বা পুরোনো content overwrite |
| `a` | Append; শেষে content যোগ |
| `x` | Create only; file আগে থাকলে error |
| `b` | Binary mode modifier |
| `t` | Text mode modifier; default |
| `r+` | Read and write; file থাকতে হবে |
| `w+` | Read and write; পুরোনো content overwrite |
| `a+` | Read and append |

Mode modifier একসঙ্গে ব্যবহার করা যায়, যেমন `rb`, `wb`।

## 7. Read the Entire File with `read()`

## 8. Read One Line with `readline()`

`readline()` একবারে একটি line পড়ে। বারবার call করলে পরের line পাওয়া যায়।

````python
with open("part7_first_file.txt", "r", encoding="utf-8") as file:
    line1 = file.readline()
    line2 = file.readline()

print("Line 1:", line1, end="")
print("Line 2:", line2, end="")
````

## 9. Read All Lines with `readlines()`

`readlines()` সব line-কে একটি list হিসেবে return করে। প্রতিটি line-এ newline character থাকতে পারে।

````python
with open("part7_first_file.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print(lines)
print("Number of lines:", len(lines))
````

## 10. Loop Through a File

বড় file-এর ক্ষেত্রে পুরো content একবারে memory-তে না নিয়ে line by line loop করা সুবিধাজনক।

````python
with open("part7_first_file.txt", "r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        print(f"{line_number}: {line.rstrip()}")
````

## 13. Create a File Safely with `x`

`x` mode কেবল নতুন file তৈরি করে। একই নামে file আগে থাকলে `FileExistsError` হয়।

````python
try:
    with open("part7_exclusive_create.txt", "x", encoding="utf-8") as file:
        file.write("This file was created using x mode.\n")
    print("New file created.")
except FileExistsError:
    print("The file already exists; it was not overwritten.")
````

## 14. File Pointer: `tell()` and `seek()`

File-এর read/write position-কে file pointer বলা যায়।
- `tell()` current position জানায়।
- `seek(position)` নির্দিষ্ট position-এ নিয়ে যায়।

````python
with open("part7_first_file.txt", "r", encoding="utf-8") as file:
    print("Initial position:", file.tell())

    first_part = file.read(5)
    print("Read:", first_part)
    print("Position now:", file.tell())

    file.seek(0)
    print("After seek:", file.tell())
    print("Read again:", file.read(5))
````

````python
bangla_text = "আমি Python শিখছি।"

with open("part7_bangla.txt", "w", encoding="utf-8") as file:
    file.write(bangla_text)

with open("part7_bangla.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

# Section C — CSV Files

## 21. What Is a CSV File?

CSV-এর পূর্ণরূপ Comma-Separated Values। এটি tabular data text format-এ রাখে।

উদাহরণ:
```text
student_id,name,marks
101,Amina,85
102,Rahim,78
103,Nila,92
```

Python-এর built-in `csv` module দিয়ে CSV read ও write করা যায়।

## 22. Write a CSV File with `csv.writer()`

`newline=""` ব্যবহার করলে Windows-এ অতিরিক্ত blank line হওয়ার সমস্যা এড়ানো যায়।

````python
import csv

rows = [
    ["student_id", "name", "marks"],
    [101, "Amina", 85],
    [102, "Rahim", 78],
    [103, "Nila", 92]
]

with open("part7_students.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(rows)

print("CSV file created.")
````

## 23. Read a CSV File with `csv.reader()`

````python
import csv

with open("part7_students.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
````

## 24. Write CSV Using `DictWriter`

`DictWriter` dictionary-এর key-গুলোকে column header হিসেবে ব্যবহার করতে পারে।

````python
import csv

students = [
    {"student_id": 201, "name": "Hasan", "marks": 88},
    {"student_id": 202, "name": "Mina", "marks": 91},
    {"student_id": 203, "name": "Rafi", "marks": 76}
]

fieldnames = ["student_id", "name", "marks"]

with open("part7_students_dict.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(students)

print("Dictionary-based CSV created.")
````

## 25. Read CSV Using `DictReader`

````python
import csv

with open("part7_students_dict.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["name"], "scored", row["marks"])
````

# Section D — JSON Files

## 26. What Is JSON?

JSON-এর পূর্ণরূপ JavaScript Object Notation। এটি structured data সংরক্ষণ ও আদান-প্রদানের বহুল ব্যবহৃত format।

Python `json` module-এর গুরুত্বপূর্ণ functions:
- `json.dump()` — Python object file-এ লেখে।
- `json.load()` — JSON file পড়ে Python object তৈরি করে।
- `json.dumps()` — Python object থেকে JSON string তৈরি করে।
- `json.loads()` — JSON string থেকে Python object তৈরি করে।

## 27. Write JSON with `json.dump()`

````python
import json

student = {
    "student_id": 101,
    "name": "Amina",
    "department": "CSE",
    "cgpa": 3.85,
    "subjects": ["Python", "Database", "Statistics"]
}

with open("part7_student.json", "w", encoding="utf-8") as file:
    json.dump(student, file, indent=4, ensure_ascii=False)

print("JSON file created.")
````

## 28. Read JSON with `json.load()`

````python
import json

with open("part7_student.json", "r", encoding="utf-8") as file:
    loaded_student = json.load(file)

print(loaded_student)
print("Name:", loaded_student["name"])
print("Subjects:", loaded_student["subjects"])
````

## 29. Convert Between Python Objects and JSON Strings

````python
import json

data = {"name": "Rahim", "active": True, "score": 88}

json_text = json.dumps(data, indent=2)
print(json_text)

python_data = json.loads(json_text)
print(python_data)
print(type(python_data))
````

# Section E — Exception Handling

## 31. What Is an Exception?

Program execution-এর সময় যে error ঘটলে normal flow interrupt হয়, তাকে exception বলা হয়।

কিছু common exception:
- `ZeroDivisionError` — শূন্য দিয়ে ভাগ
- `ValueError` — value-এর type ঠিক হলেও value invalid
- `TypeError` — incompatible type operation
- `NameError` — undefined name
- `IndexError` — sequence-এর invalid index
- `KeyError` — dictionary key নেই
- `FileNotFoundError` — file পাওয়া যায়নি
- `PermissionError` — permission নেই
- `ModuleNotFoundError` — module পাওয়া যায়নি

Exception handling program-কে predictable way-তে error সামলাতে সাহায্য করে।

## 32. First `try` and `except`

`try` block-এ এমন code রাখা হয় যেখানে exception হতে পারে। Matching `except` block exception handle করে।

````python
try:
    number = int(input("Enter an integer: "))
    print("You entered:", number)
except ValueError:
    print("Invalid input. Please enter a whole number.")
````

## 33. Handle Division by Zero

````python
try:
    numerator = float(input("Numerator: "))
    denominator = float(input("Denominator: "))

    result = numerator / denominator
    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
except ValueError:
    print("Please enter valid numeric values.")
````

## 34. Multiple `except` Blocks

Different exception-এর জন্য আলাদা handling করা যায়। Specific exception আগে এবং broader exception পরে রাখবে।

````python
values = [10, 20, 30]

try:
    index = int(input("Enter an index (0-2): "))
    print(values[index])
except ValueError:
    print("Index must be an integer.")
except IndexError:
    print("That index does not exist.")
````

## 35. `else` Block

`else` block কেবল তখন execute হয় যখন `try` block-এ কোনো exception ঘটে না।

````python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid integer.")
else:
    print("Conversion successful.")
    print("Double:", number * 2)
````

## 36. `finally` Block

`finally` block exception হোক বা না হোক, সাধারণত execute হয়। Cleanup operation-এর জন্য এটি কাজে লাগে।

````python
try:
    print("Trying an operation.")
    result = 10 / 2
    print("Result:", result)
except ZeroDivisionError:
    print("Division error.")
finally:
    print("This cleanup section runs at the end.")
````

## 37. Full Structure: `try-except-else-finally`

````python
try:
    value = int(input("Enter an integer: "))
    result = 100 / value
except ValueError:
    print("Please enter an integer.")
except ZeroDivisionError:
    print("Zero is not allowed.")
else:
    print("Calculation successful:", result)
finally:
    print("Program section finished.")
````

## 38. Catching an Exception Object

`as error` দিয়ে exception object পাওয়া যায়। Debugging-এর সময় message দেখানো কাজে লাগে। User-facing program-এ sensitive internal details প্রকাশ করা উচিত নয়।

````python
try:
    number = int("abc")
except ValueError as error:
    print("Error type:", type(error).__name__)
    print("Error message:", error)
````

## 39. Avoid Bare `except`

`except:` সব ধরনের exception ধরতে পারে, এমনকি KeyboardInterrupt-এর মতো control-flow exception-ও। সাধারণত এটি এড়িয়ে specific exception ধরবে।

ভালো:
```python
try:
    number = int(user_text)
except ValueError:
    print("Please enter a valid integer.")
```

কম ভালো:
```python
try:
    number = int(user_text)
except:
    print("Something went wrong.")
```

## 40. Raise an Exception Manually

`raise` ব্যবহার করে নিজের code থেকে exception তৈরি করা যায়। Invalid input বা violated rule জানাতে এটি কাজে লাগে।

````python
def calculate_discount(price, discount_percent):
    if price < 0:
        raise ValueError("Price cannot be negative.")

    if not 0 <= discount_percent <= 100:
        raise ValueError("Discount must be between 0 and 100.")

    return price * (1 - discount_percent / 100)

print(calculate_discount(1000, 10))
````

## 41. Handle a Manually Raised Exception

````python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    return age

try:
    print(set_age(-5))
except ValueError as error:
    print("Invalid age:", error)
````

## 42. Custom Exception Class

নিজের application-এর জন্য meaningful exception class তৈরি করা যায়। সাধারণত `Exception` inherit করা হয়।

````python
class InsufficientBalanceError(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError("Not enough balance.")
    return balance - amount

try:
    remaining = withdraw(1000, 1500)
    print("Remaining:", remaining)
except InsufficientBalanceError as error:
    print("Withdrawal failed:", error)
````

## 43. Exception Handling with File Operations

File না থাকলে `FileNotFoundError` handle করা যায়। এখানে file create করা হচ্ছে না; শুধু missing-file situation দেখানো হচ্ছে।

````python
from pathlib import Path

missing_file = Path("this_file_does_not_exist_part7.txt")

try:
    content = missing_file.read_text(encoding="utf-8")
    print(content)
except FileNotFoundError:
    print("The requested file was not found.")
````

এই project-এ user note লিখবে, file-এ save হবে এবং পরে read করা যাবে।

## 45. Mini Project 2 — Student CSV Report

CSV file থেকে student marks পড়বে এবং average বের করবে। CSV file আগের section-এ তৈরি করা হয়েছে।

````python
import csv

marks = []

try:
    with open("part7_students.csv", "r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for row in reader:
            marks.append(float(row["marks"]))

    if marks:
        average = sum(marks) / len(marks)
        print("Students:", len(marks))
        print("Average marks:", round(average, 2))
    else:
        print("No student records found.")

except FileNotFoundError:
    print("CSV file not found. Run the CSV creation cell first.")
except (ValueError, KeyError) as error:
    print("The CSV data is not in the expected format:", error)
````

## 46. Mini Project 3 — JSON Contact Book

Contact information JSON file-এ save ও read করবে। এই example একটি small practice contact book তৈরি করে।

````python
import json
from pathlib import Path

contacts_path = Path("part7_contacts.json")

new_contact = {
    "name": input("Contact name: "),
    "phone": input("Phone number: ")
}

try:
    if contacts_path.exists():
        contacts = json.loads(contacts_path.read_text(encoding="utf-8"))
    else:
        contacts = []

    contacts.append(new_contact)

    contacts_path.write_text(
        json.dumps(contacts, indent=4, ensure_ascii=False),
        encoding="utf-8"
    )

    print("Contact saved.")
    print(json.dumps(contacts, indent=4, ensure_ascii=False))

except json.JSONDecodeError:
    print("The existing contacts file contains invalid JSON.")
````

## 47. Mini Project 4 — Safe Calculator with Exception Handling

User input validate করে calculation করবে।

````python
def safe_calculator():
    try:
        first = float(input("First number: "))
        operator = input("Operator (+, -, *, /): ")
        second = float(input("Second number: "))

        if operator == "+":
            answer = first + second
        elif operator == "-":
            answer = first - second
        elif operator == "*":
            answer = first * second
        elif operator == "/":
            answer = first / second
        else:
            print("Unknown operator.")
            return

        print("Answer:", answer)

    except ValueError:
        print("Please enter valid numbers.")
    except ZeroDivisionError:
        print("Cannot divide by zero.")

safe_calculator()
````

### Solutions 1–5: Text files and paths

````python
from pathlib import Path

# 1. Create and write
intro_path = Path("my_intro_part7.txt")
intro_path.write_text(
    "My name is Amina.\nI am learning Python.\n",
    encoding="utf-8"
)

# 2. Read
print(intro_path.read_text(encoding="utf-8"))

# 3. Append
with intro_path.open("a", encoding="utf-8") as file:
    file.write("I am practising file handling.\n")

# 4. Count lines
with intro_path.open("r", encoding="utf-8") as file:
    lines = file.readlines()
print("Line count:", len(lines))

# 5. Check existence
print("File exists:", intro_path.exists())
````

### Solutions 6–8: CSV and JSON

````python
import csv
import json

# 6. Write CSV
student_rows = [
    {"id": 1, "name": "Amina", "marks": 85},
    {"id": 2, "name": "Rahim", "marks": 78},
    {"id": 3, "name": "Nila", "marks": 92}
]

with open("practice_students.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["id", "name", "marks"])
    writer.writeheader()
    writer.writerows(student_rows)

# 7. Read CSV
with open("practice_students.csv", "r", newline="", encoding="utf-8") as file:
    for row in csv.DictReader(file):
        print(row["name"])

# 8. Write and read JSON
profile = {"name": "Amina", "department": "CSE", "active": True}

with open("practice_profile.json", "w", encoding="utf-8") as file:
    json.dump(profile, file, indent=4)

with open("practice_profile.json", "r", encoding="utf-8") as file:
    loaded_profile = json.load(file)

print(loaded_profile)
````

### Solutions 9–15: Exception handling

````python
# 9. Integer input validation
try:
    value = int(input("Enter an integer: "))
    print(value)
except ValueError:
    print("Invalid integer.")

# 10. Division by zero
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Cannot divide by zero.")

# 11. List index validation
items = ["red", "green", "blue"]
try:
    index = int(input("Enter index 0-2: "))
    print(items[index])
except ValueError:
    print("Index must be an integer.")
except IndexError:
    print("Index is outside the list.")

# 12. Raise ValueError
def positive_only(number):
    if number < 0:
        raise ValueError("Number must not be negative.")
    return number

try:
    print(positive_only(-2))
except ValueError as error:
    print(error)

# 13. Custom exception
class InvalidAgeError(Exception):
    pass

def validate_age(age):
    if age < 0:
        raise InvalidAgeError("Age cannot be negative.")
    return age

try:
    print(validate_age(-1))
except InvalidAgeError as error:
    print(error)

# 14. Missing file
try:
    with open("missing_practice_file.txt", "r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("File does not exist.")

# 15. try-except-else-finally
try:
    result = int("25") * 2
except ValueError:
    print("Conversion failed.")
else:
    print("Result:", result)
finally:
    print("Finished.")
````

## 48. Common Errors and Troubleshooting

| Error / Problem | কারণ | সমাধান |
|---|---|---|
| `FileNotFoundError` | File path ভুল বা file নেই | `Path.cwd()` ও path যাচাই |
| `PermissionError` | File/folder access permission নেই | Permission ও file usage পরীক্ষা |
| `IsADirectoryError` | File হিসেবে directory open করা হয়েছে | সঠিক file path দাও |
| `FileExistsError` | `x` mode-এ file আগে থেকেই আছে | অন্য নাম ব্যবহার বা existence check |
| `UnicodeDecodeError` | ভুল encoding দিয়ে read | সঠিক encoding, সাধারণত UTF-8, যাচাই |
| `JSONDecodeError` | JSON syntax invalid বা empty file | JSON content validate |
| `csv.Error` | CSV format/parsing issue | delimiter ও file structure পরীক্ষা |
| `ValueError` | Input expected format নয় | Input validation |
| `ZeroDivisionError` | denominator শূন্য | ভাগের আগে পরীক্ষা বা exception handle |

### Good practices
- File open করার সময় `with` ব্যবহার করো।
- Text file-এর encoding explicit দাও।
- `w` mode ব্যবহারের আগে overwrite হবে কি না যাচাই করো।
- User input validate করো।
- Specific exception ধরো; bare `except` এড়িয়ে চলো।
- Error হলে প্রয়োজনীয় message দাও; নীরবে ignore কোরো না।
