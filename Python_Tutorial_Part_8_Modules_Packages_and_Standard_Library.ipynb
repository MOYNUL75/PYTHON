## 1. What is a Module?

A **module** is a Python file (`.py`) that contains reusable code, such as variables, functions and classes.

সহজভাবে: একটি Python file-এর মধ্যে function বা variable লিখে রাখলে, অন্য Python file থেকে সেটি import করে ব্যবহার করা যায়। এতে একই code বারবার লিখতে হয় না।

**Example:** Python-এর `math` module-এ mathematical functions আছে।

````python
import math

print(math.sqrt(25))
print(math.pi)
print(math.factorial(5))
````

**Expected output**
```text
5.0
3.141592653589793
120
```

- `import math` — `math` module যুক্ত করে।
- `math.sqrt(25)` — ২৫-এর square root বের করে।
- `math.pi` — pi-এর মান দেয়।
- `math.factorial(5)` — ৫! হিসাব করে।

## 2. Why Do We Use Modules?

Modules help us:
- Reuse code instead of writing it repeatedly.
- Organize a large program into smaller files.
- Make programs easier to test and maintain.
- Use ready-made Python functionality.

**Real-life example:** একটি student management system-এ student-related functions `students.py`-তে এবং result-related functions `results.py`-তে রাখা যেতে পারে।

## 3. Importing a Module

### 3.1 Import the whole module

**Syntax**
```python
import module_name
```

````python
import math

number = 49
result = math.sqrt(number)

print("Square root:", result)
````

**Expected output**
```text
Square root: 7.0
```
`math.sqrt()` লেখার সময় module-এর নামের পরে dot (`.`) ব্যবহার করা হয়।

### 3.2 Import a module with an alias

`as` দিয়ে module-এর ছোট নাম দেওয়া যায়।

````python
import math as m

print(m.sqrt(81))
print(m.ceil(4.2))
print(m.floor(4.8))
````

**Expected output**
```text
9.0
5
4
```
`m` হলো `math` module-এর alias।

### 3.3 Import specific items

`from ... import ...` দিয়ে module-এর নির্দিষ্ট function বা variable import করা যায়।

````python
from math import sqrt, pi

print(sqrt(64))
print(pi)
````

**Expected output**
```text
8.0
3.141592653589793
```
এখানে `math.sqrt()` না লিখে সরাসরি `sqrt()` ব্যবহার করা হয়েছে।

### 3.4 Import a specific item with an alias

````python
from math import factorial as fact

print(fact(6))
````

**Expected output**
```text
720
```

### 3.5 Avoid `from module import *` in normal projects

এভাবে সব নাম import করা যায়:
```python
from math import *
```
কিন্তু এতে কোন নাম কোন module থেকে এসেছে তা বোঝা কঠিন হতে পারে এবং একই নামের conflict হতে পারে। তাই সাধারণত পুরো module বা নির্দিষ্ট item import করা ভালো।

### 5.1 `math` — Mathematical functions

````python
import math

print("Square root:", math.sqrt(36))
print("Power:", math.pow(2, 3))
print("Ceiling:", math.ceil(5.1))
print("Floor:", math.floor(5.9))
print("GCD:", math.gcd(12, 18))
````

### 5.2 `random` — Random values

`random` module random number বা random selection তৈরি করতে সাহায্য করে। Security-sensitive password বা token তৈরির জন্য `random` নয়, `secrets` module ব্যবহার করতে হয়।

````python
import random

print("Random integer:", random.randint(1, 10))
print("Random decimal:", random.random())

colors = ["red", "green", "blue"]
print("Random color:", random.choice(colors))
````

**Note:** Output প্রতিবার পরিবর্তিত হতে পারে, কারণ value randomভাবে তৈরি হয়।

Common functions:
- `randint(a, b)` — a থেকে b পর্যন্ত integer (দুই প্রান্তসহ)।
- `random()` — 0.0 থেকে 1.0-এর কম একটি float।
- `choice(sequence)` — sequence থেকে একটি item নির্বাচন করে।
- `shuffle(list)` — list-কে in-place এলোমেলো করে।

````python
import random

numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(numbers)  # Order may change
````

### 5.7 `json` — JSON data

JSON (JavaScript Object Notation) data exchange-এর একটি common format। Python dictionary এবং JSON object দেখতে কাছাকাছি হলেও এক জিনিস নয়।

## 6. Creating Your Own Python Module

এখন আমরা নিজেরা একটি module তৈরি করব।

ধরা যাক, `calculator.py` নামে একটি file থাকবে, যেখানে calculation functions থাকবে। অন্য file বা notebook থেকে সেটি import করব।

````python
from pathlib import Path

module_code = (
    "def add(a, b):\n"
    "    return a + b\n\n"
    "def subtract(a, b):\n"
    "    return a - b\n\n"
    "def multiply(a, b):\n"
    "    return a * b\n"
)

Path("calculator.py").write_text(module_code, encoding="utf-8")
print("calculator.py created")
````

এখন তৈরি করা module import করি। যদি একই notebook-এ module file আগে import করা হয়ে থাকে, file পরিবর্তনের পর `importlib.reload()` কাজে লাগতে পারে।

````python
import calculator

print(calculator.add(10, 5))
print(calculator.subtract(10, 5))
print(calculator.multiply(10, 5))
````

**Expected output**
```text
15
5
50
```

**Important:** `calculator.py` এবং notebook একই folder-এ থাকলে সাধারণত সরাসরি import করা যায়। অন্য folder হলে Python-এর module search path বা package structure বুঝতে হবে।

## 7. Creating a Module with Variables and a Class

Module-এ শুধু function নয়, variable ও class-ও থাকতে পারে।

````python
from pathlib import Path

module_code = (
    'COURSE_NAME = "Python Programming"\n\n'
    'def welcome(name):\n'
    '    return f"Welcome, {name}!"\n\n'
    'class Student:\n'
    '    def __init__(self, name):\n'
    '        self.name = name\n\n'
    '    def introduce(self):\n'
    '        return f"My name is {self.name}."\n'
)

Path("course_info.py").write_text(module_code, encoding="utf-8")
print("course_info.py created")
````

````python
import course_info

print(course_info.COURSE_NAME)
print(course_info.welcome("Nila"))

student = course_info.Student("Rahim")
print(student.introduce())
````

**Expected output**
```text
Python Programming
Welcome, Nila!
My name is Rahim.
```

## 8. What is a Package?

A **package** is a way to organize related Python modules in directories. A package can contain multiple modules and subpackages.

Example structure:
```text
my_package/
    __init__.py
    calculator.py
    greetings.py
```

- `my_package` — package folder.
- `__init__.py` — package initialization file; it may be empty.
- `calculator.py` — one module.
- `greetings.py` — another module.

Modern Python can also use namespace packages without `__init__.py`, but for learning, we will use a regular package with `__init__.py`.

## 9. Creating Your Own Package

The next cell creates a small package with two modules.

````python
from pathlib import Path

package = Path("my_package")
package.mkdir(exist_ok=True)

(package / "__init__.py").write_text(
    '"""A small practice package."""\n',
    encoding="utf-8"
)

(package / "calculator.py").write_text(
    "def add(a, b):\n    return a + b\n",
    encoding="utf-8"
)

(package / "greetings.py").write_text(
    'def greet(name):\n    return f"Hello, {name}!"\n',
    encoding="utf-8"
)

print("Package created:", package.resolve())
````

### 9.1 Import from a package

````python
from my_package.calculator import add
from my_package.greetings import greet

print(add(20, 30))
print(greet("Karim"))
````

**Expected output**
```text
50
Hello, Karim!
```
**Import pattern**
```python
from package_name.module_name import function_name
```

### 9.2 Import a module from a package using an alias

````python
import my_package.calculator as calc

print(calc.add(7, 8))
````

## 14. Mini Project 2 — Student Result Report

`statistics` module দিয়ে student-এর average, highest এবং lowest marks বের করব।

````python
import statistics

student_name = "Nila"
marks = [78, 85, 92, 88, 77]

report = {
    "student": student_name,
    "average": statistics.mean(marks),
    "highest": max(marks),
    "lowest": min(marks)
}

print("Student:", report["student"])
print("Average:", report["average"])
print("Highest:", report["highest"])
print("Lowest:", report["lowest"])
````

## 16. Mini Project 4 — Save Student Data as JSON

এই project-এ student information JSON file-এ save এবং read করব।

````python
import json
from pathlib import Path

student = {
    "id": 101,
    "name": "Rahim",
    "department": "CSE",
    "cgpa": 3.80
}

file_path = Path("student_data.json")

file_path.write_text(
    json.dumps(student, indent=4),
    encoding="utf-8"
)

loaded_student = json.loads(file_path.read_text(encoding="utf-8"))

print("Saved file:", file_path.resolve())
print("Student name:", loaded_student["name"])
print("Department:", loaded_student["department"])
````

**Expected output**
```text
Student name: Rahim
Department: CSE
```
The file `student_data.json` will be created in the current working directory.

## 19. Common Errors and Troubleshooting

### 19.1 `ModuleNotFoundError`
**Meaning:** Python cannot find the module.

Possible reasons:
- The module/package is not installed.
- The `.py` file is not in the current folder or module search path.
- The module name is misspelled.

Check:
```python
import os
print(os.getcwd())
```

### 19.2 `ImportError`
Python found a module, but the requested name could not be imported. Check the function or variable name and spelling.

### 19.3 Naming conflict
নিজের file-এর নাম standard module-এর মতো রাখবে না। যেমন `random.py`, `math.py`, `json.py` নাম দিলে standard module import-এ সমস্যা হতে পারে।

### 19.4 Changes in a module are not reflected
একই running session-এ module আগে import করা থাকলে Python cached module ব্যবহার করতে পারে। Practice-এর সময়:
```python
import importlib
import calculator
importlib.reload(calculator)
```
অথবা kernel restart করে আবার import করতে পারো।

### 19.5 Package import fails
- Folder name এবং module name ঠিক আছে কি না দেখো।
- Notebook-এর current working directory যাচাই করো।
- Regular package হলে `__init__.py` আছে কি না দেখো।
- File তৈরি করার cell আগে run করেছ কি না নিশ্চিত হও।
